### Exercici 0.1
Amb `a = 10` i `b = 3`, calcula suma, resta, producte, divisió real, divisió entera, residu i potència. Mostra també el tipus de cada resultat.

````python
# Escriu aquí la teva solució
a = 10
b = 3

suma = a + b
resta = a - b
multiplicacion = a * b
division_real = a / b
division_entera = a // b
residuo = a % b
potencia = a ** b

print("Suma:", suma)
print("Resta:", resta)
print("Multiplicacion:", multiplicacion)
print("Division Real:", division_real)
print("Division Entera: ", division_entera)
print("Residuo:", residuo)
print("Pontecia:", potencia)
````
